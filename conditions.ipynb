```python
n = int(input("Enter 1st number: "))
s = int(input("Enter 2nd number: "))
m = int(input("Enter 3rd number: "))


if n > s and n > m:
    print(f"{n} is bigger")
elif s > n and s > m:
    print(f"{s} is bigger")
else:
    print(f"{m} is bigger")
```
