```python
rows = int(input("How many rows do you want?: "))

for i in range(rows):
    if i == 0 or i == rows-1:
        print("*      " * rows)
        print()
    else:
        for j in range(rows):
                if j == 0 or j == rows-1:
                    print("*",end="")
                else:
                    print("         ",end="")
        print()
        print()
    
```
